```python
# List comprehension
mylist = [i for i in range(0,10)]
mylist
```

```python
# Dictionary comprehension
my_dict = {i:i**2 for i in range(0,10)}
my_dict
```

```python
# tuple
my_tuple = tuple(i for i in range(0,5))
my_tuple
```

```python
# Collection
from collections import Counter

data = ['a', 'b', 'c', 'a', 'b', 'a']
count = Counter(data)
print(count)

s1 = 'cat'
c1 = Counter(s1)
print(c1)

s2 = 'act'
c2 = Counter(s2)
print(c2)

print(c1 == c2)
```
